### 3. Función de referencia ineficiente (calcular_indicadores_lento) y medición de tiempo

```
def calcular_indicadores_lento(df):
    """Version de referencia DELIBERADAMENTE ineficiente (no modificar)."""
    resultados = []
    for id_emp in df['id_empleado'].unique():
        filas = df[df['id_empleado'] == id_emp]
        suma_calif = 0
        suma_bono = 0
        n = 0
        for _, fila in filas.iterrows():
            suma_calif += fila['calificacion_desempeno']
            suma_bono += fila['bono_asignado']
            n += 1
        resultados.append({
            'id_empleado': id_emp,
            'calificacion_promedio': suma_calif / n,
            'bono_total': suma_bono
        })
    return pd.DataFrame(resultados)

# Mida el tiempo de ejecucion aqui (use una muestra si es necesario)
```
