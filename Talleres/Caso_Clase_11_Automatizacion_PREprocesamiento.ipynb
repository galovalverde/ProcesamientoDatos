## Caso: Pipeline automatizado de preprocesamiento para la fintech "PagaYa"

**Contexto**

PagaYa es una fintech que recibe, cada día, nuevos lotes de datos de transacciones provenientes de distintos comercios afiliados. Hasta ahora, cada vez que llega un archivo nuevo, un analista debe limpiar, transformar y validar manualmente los datos antes de que puedan usarse en los modelos de scoring de riesgo. Esta forma de trabajar es lenta y propensa a errores humanos, por lo que el equipo de ingeniería de datos le solicita construir un **pipeline de preprocesamiento automatizado y reutilizable**, capaz de aplicarse a cualquier archivo nuevo con la misma estructura sin necesidad de reescribir código cada vez.

**Insumos**

Se le entregan dos archivos de transacciones (`transacciones_lote1.csv` y `transacciones_lote2.csv`) con la misma estructura de columnas (`id_transaccion`, `comercio`, `fecha`, `monto`, `categoria_comercio`, `edad_cliente`, `metodo_pago`, `es_fraude`), pero con diferencias entre lotes propias de un entorno real: distinto porcentaje y patrón de valores nulos, categorías con variantes de escritura distintas entre lotes, y rangos de outliers distintos en `monto` y `edad_cliente`. Un tercer archivo (`config_preprocesamiento.json`) define de forma declarativa qué columnas son numéricas, cuáles categóricas, qué estrategia de imputación usar por columna y los umbrales para detección de outliers.

**Requerimientos técnicos**

- Diseñar el pipeline como **funciones o clases reutilizables** (no código repetido "copiado y pegado" para cada lote), de forma que agregar un tercer lote no requiera modificar la lógica interna.
- Construir el flujo utilizando `Pipeline` y `ColumnTransformer` de scikit-learn (o una estructura equivalente propia), separando claramente los pasos de imputación, codificación de categóricas, escalado de numéricas y tratamiento de outliers.
- Leer las reglas de imputación, columnas numéricas/categóricas y umbrales de outliers **desde el archivo de configuración JSON**, evitando "hardcodear" nombres de columnas dentro de la lógica del pipeline.
- Implementar al menos un **transformador personalizado** (custom transformer, heredando de `BaseEstimator`/`TransformerMixin` o equivalente) para una regla de negocio específica (por ejemplo, estandarizar las categorías de comercio antes de codificar).
- Incorporar una función de **validación de esquema** que verifique, antes de procesar cualquier archivo nuevo, que contiene las columnas esperadas y tipos de dato compatibles, reportando un error claro si no es así.
- Aplicar el mismo pipeline, sin modificar su lógica interna, a los dos lotes de transacciones y demostrar que produce resultados consistentes y comparables entre ambos.
- Registrar mediante *logging* (o un mecanismo equivalente de trazabilidad) cada paso ejecutado y las métricas relevantes (filas afectadas por imputación, outliers tratados, etc.) para cada lote procesado.

**Entregables**

Notebook `.ipynb` con: definición del pipeline reutilizable y sus componentes (funciones/clases/transformador personalizado), lectura y aplicación de la configuración externa, validación de esquema, ejecución automatizada sobre ambos lotes con su respectivo log de trazabilidad, y una sección final de conclusiones que explique qué cambiaría si mañana llega un tercer lote con una columna adicional.

**Criterios de evaluación**

- Diseño modular y reutilizable del pipeline (sin duplicación de código entre lotes) (30%)
- Uso correcto de `Pipeline`/`ColumnTransformer` y de al menos un transformador personalizado (25%)
- Uso efectivo del archivo de configuración externo para parametrizar el proceso (20%)
- Validación de esquema y manejo de errores (15%)
- Trazabilidad (logging) y claridad de las conclusiones (10%)
