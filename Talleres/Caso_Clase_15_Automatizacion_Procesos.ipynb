## Caso: Automatización del proceso ETL de RR.HH. para "Grupo Empresarial Andino"

**Contexto**

En el caso anterior (Clase 13-14) usted diseñó e implementó, de forma manual y "de una sola vez", el proceso ETL que consolida `empleados`, `departamentos` y `evaluaciones` en una base de datos SQLite. El proceso funcionó bien, pero la gerencia de Recursos Humanos ahora quiere ejecutarlo **cada mes** sin que un analista deba correr el notebook celda por celda ni preocuparse por duplicar información si el proceso se ejecuta más de una vez por error. Le solicitan **automatizar** ese pipeline para que sea confiable, incremental y capaz de recuperarse de errores.

**Insumos**

Se le entrega la base de datos ya cargada del caso anterior (`grupo_andino.db`) y dos archivos que simulan la llegada de un **nuevo mes de operación**: `empleados_nuevos_mes.csv` (contratos nuevos y algunas actualizaciones de `salario_base` de empleados ya existentes) y `evaluaciones_nuevo_trimestre.csv` (evaluaciones del trimestre más reciente). Ambos archivos contienen, de forma intencional, algunos registros que ya existen en la base de datos, para que su proceso deba decidir correctamente si insertarlos o ignorarlos/actualizarlos.

**Requerimientos técnicos**

- Diseñar el proceso como una **función orquestadora única** (`ejecutar_pipeline_mensual()` o similar) que encapsule las etapas de extracción, transformación y carga, de forma que ejecutarla sea el único paso manual necesario cada mes.
- Implementar una **carga incremental** (no un `TRUNCATE`/recarga completa): los empleados nuevos deben insertarse, los empleados existentes con `salario_base` distinto deben actualizarse, y las evaluaciones nuevas deben agregarse sin duplicar las que ya existían.
- Garantizar **idempotencia**: si la función se ejecuta dos veces seguidas sobre los mismos archivos de entrada, la base de datos debe quedar exactamente igual que si se hubiera ejecutado una sola vez (sin duplicados ni datos corruptos).
- Incluir manejo de errores con `try/except` en cada etapa crítica (lectura de archivo, conexión a la base, escritura), de forma que si una etapa falla, el proceso lo reporte claramente y no deje la base de datos en un estado parcial o inconsistente.
- Registrar cada ejecución en una **tabla de control** dentro de la misma base de datos (por ejemplo `log_ejecuciones`, con fecha de ejecución, cantidad de registros insertados/actualizados por tabla, y si terminó con éxito o con error).
- Simular al menos dos ejecuciones consecutivas del pipeline (una con los archivos del "nuevo mes" y otra repitiendo la misma ejecución) y demostrar, con evidencia (conteos de filas, consultas SQL), que la segunda ejecución no alteró los datos ya cargados.

**Entregables**

Notebook `.ipynb` con: la función orquestadora completa, la lógica de carga incremental e idempotente, el manejo de errores, la tabla de control de ejecuciones, la simulación de las dos ejecuciones con su evidencia de idempotencia, y una sección final de conclusiones que explique qué mecanismo agregaría si el proceso tuviera que ejecutarse automáticamente sin supervisión humana (por ejemplo, notificaciones o reintentos).

**Criterios de evaluación**

- Diseño de la función orquestadora y encapsulamiento del pipeline (20%)
- Correcta implementación de la carga incremental (insertar/actualizar sin duplicar) (30%)
- Demostración efectiva de idempotencia ante ejecuciones repetidas (25%)
- Manejo de errores y tabla de control de ejecuciones (15%)
- Claridad del notebook y conclusiones (10%)
