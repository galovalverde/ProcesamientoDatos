## Caso: Preparación de datos de viajes para "EcoMovil", servicio de bicicletas y patinetas compartidas

**Contexto**

EcoMovil es una empresa de movilidad urbana que opera bicicletas y patinetas eléctricas compartidas en Guayaquil. El área de analítica ha recibido el archivo `viajes_ecomovil.csv` con el registro de viajes del último mes, y necesita dejarlo listo para alimentar un modelo de predicción de demanda. Antes de eso, le solicitan a usted limpiar el dataset, transformarlo a un formato adecuado para análisis/modelado, y detectar los viajes atípicos que podrían distorsionar los resultados.

**Descripción del dataset**

El archivo contiene, entre otras, las columnas: `id_viaje`, `tipo_vehiculo` (bicicleta/patineta), `fecha_inicio`, `hora_inicio`, `duracion_minutos`, `distancia_km`, `costo`, `zona`, `edad_usuario` y `metodo_pago`. El dataset presenta: valores faltantes en `edad_usuario` y `costo`, registros duplicados, texto con espacios/mayúsculas inconsistentes en `zona` y `metodo_pago`, viajes con `duracion_minutos` o `distancia_km` negativos o en cero (errores de sensor), y un número reducido de viajes con duraciones o costos extremadamente altos.

**Parte 1 — Limpieza de Datos (Clase 08)**

- Detectar y eliminar duplicados.
- Identificar valores faltantes por columna y aplicar una estrategia de tratamiento justificada (imputación o eliminación, según la variable).
- Corregir valores imposibles (`duracion_minutos` o `distancia_km` ≤ 0) considerando reglas de negocio razonables.
- Normalizar el texto de las columnas categóricas (`zona`, `metodo_pago`, `tipo_vehiculo`): espacios, mayúsculas/minúsculas y posibles variantes de escritura.

**Parte 2 — Transformación de Datos (Clase 09)**

- Extraer nuevas variables a partir de `fecha_inicio` y `hora_inicio`: día de la semana, si es fin de semana, y franja horaria (mañana/tarde/noche).
- Calcular una variable derivada `velocidad_promedio` (distancia_km / duracion_minutos en horas).
- Codificar las variables categóricas (`tipo_vehiculo`, `zona`, `metodo_pago`) usando una técnica adecuada (one-hot encoding o label encoding, justificando la elección).
- Escalar o normalizar las variables numéricas relevantes (`duracion_minutos`, `distancia_km`, `costo`) usando una técnica de estandarización o normalización min-max.

**Parte 3 — Detección de Valores Atípicos (Clase 10)**

- Detectar outliers en `duracion_minutos`, `distancia_km` y `costo` aplicando al menos dos métodos distintos (por ejemplo, rango intercuartílico/IQR y z-score), comparando los resultados.
- Visualizar los outliers mediante boxplots y/o scatter plots.
- Decidir y justificar el tratamiento de los outliers detectados (eliminación, capping/winsorización, o conservarlos con marca especial), sin eliminar información válida de forma injustificada.

**Entregables**

Notebook `.ipynb` con: carga de datos, limpieza documentada (Parte 1), transformación y creación de variables (Parte 2), detección y tratamiento de outliers con visualizaciones (Parte 3), y una sección final de conclusiones sobre el impacto de cada etapa en la calidad final del dataset.

**Criterios de evaluación**

- Correcta limpieza de duplicados, nulos y valores imposibles (25%)
- Calidad y pertinencia de las transformaciones y variables derivadas (25%)
- Aplicación correcta de al menos dos métodos de detección de outliers y su comparación (30%)
- Justificación de las decisiones tomadas en cada parte (10%)
- Claridad del notebook y conclusiones (10%)
