## Caso: Consolidación de clientes para la cooperativa financiera "CoopAndina"

**Contexto**

CoopAndina es una cooperativa de ahorro y crédito que gestiona la información de sus socios en tres sistemas independientes que nunca se diseñaron para conversar entre sí: el **core bancario** (datos básicos del socio), el **CRM comercial** (datos de contacto y gestión) y la **plataforma de créditos en línea** (solicitudes y scoring). La gerencia necesita una vista única y confiable de cada socio, por lo que le encargan integrar estas tres fuentes, resolver los conflictos que surjan entre ellas y documentar claramente el origen de cada dato en la vista final.

**Fuentes de datos**

1. **`core_bancario.csv`**: un registro por socio, con `cedula`, `nombre_completo`, `fecha_nacimiento`, `sucursal` y `fecha_afiliacion`. Es la fuente considerada más confiable para los datos de identidad.
2. **`crm_comercial.xlsx`**: un registro por socio (identificado por `cedula`), con `telefono`, `correo`, `direccion` y `ultimo_contacto`. Estos datos son ingresados manualmente por asesores, por lo que contienen inconsistencias de formato y algunos valores desactualizados o duplicados.
3. **`plataforma_creditos.json`**: solicitudes de crédito por socio (una lista por `cedula`, ya que un socio puede tener varias solicitudes), con `monto_solicitado`, `estado_solicitud` y `score_crediticio`.

Las tres fuentes no siempre comparten la misma cantidad de socios (hay socios sin CRM, socios sin créditos, e incluso alguna `cedula` con formato distinto entre sistemas —con y sin guiones—).

**Parte 1 — Consolidación de fuentes (3.1.1)**

- Cargar las tres fuentes y homologar la clave de integración (`cedula`) para que coincida entre sistemas.
- Consolidar la información en una tabla única por socio, decidiendo y justificando qué tipo de combinación usar (`inner`, `left`, `outer`) según qué se quiere lograr: una vista completa de socios o solo de socios activos en las tres fuentes.
- Para los socios con múltiples solicitudes de crédito, decidir y justificar cómo resumir esa información en la vista consolidada (por ejemplo, número total de solicitudes, monto total, score más reciente).

**Parte 2 — Calidad en la integración (3.1.2)**

- Detectar socios que aparecen en más de una fuente con datos **conflictivos** (por ejemplo, distinto `nombre_completo` o `telefono` registrado en dos sistemas) y definir una regla de resolución de conflictos (por ejemplo, priorizar la fuente más confiable, o la más reciente según `ultimo_contacto`).
- Identificar y reportar los socios que quedaron con información incompleta tras la integración (ej. sin CRM o sin créditos) y decidir cómo tratarlos en la tabla final.
- Calcular al menos dos indicadores de calidad de la integración: por ejemplo, porcentaje de socios con coincidencia en las tres fuentes, y porcentaje de registros con al menos un conflicto detectado.

**Parte 3 — Gestión de metadatos (3.1.3)**

- Construir un **diccionario de metadatos** (puede ser un DataFrame o diccionario estructurado) que documente, para cada columna de la tabla consolidada: su fuente de origen, tipo de dato, si fue transformada durante la integración, y su nivel de confiabilidad.
- Registrar la **trazabilidad** del proceso: cuántos registros aportó cada fuente, cuántos se combinaron exitosamente y cuántos quedaron parciales, a modo de bitácora del proceso de integración.

**Entregables**

Notebook `.ipynb` con: carga y homologación de las tres fuentes (Parte 1), tabla consolidada con reglas de combinación justificadas, detección y resolución de conflictos con indicadores de calidad (Parte 2), diccionario de metadatos y bitácora de trazabilidad (Parte 3), y una sección final de conclusiones sobre los principales riesgos de calidad encontrados al integrar estas tres fuentes.
