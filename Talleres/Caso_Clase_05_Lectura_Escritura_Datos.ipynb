## Caso: Consolidación de datos multi-formato para "Andes Retail S.A."

**Contexto**

Andes Retail S.A., una cadena de tiendas minoristas con presencia en Guayaquil, Quito y Cuenca, ha contratado a su equipo de análisis de datos para construir un módulo de ingesta que consolide información proveniente de fuentes heterogéneas antes de que dicha información llegue a la bodega de datos corporativa. Cada sucursal y sistema interno exporta su información en un formato distinto, con inconsistencias propias de un entorno real de producción, y la gerencia de Business Intelligence necesita un pipeline reproducible que lea, valide, transforme y reescriba esta información en formatos estandarizados para su consumo posterior.

**Fuentes de datos que usted debe procesar**

1. **Ventas diarias (CSV):** archivo `ventas_sucursales.csv` separado por punto y coma (`;`), codificado en `latin-1`, con la columna de fecha en formato `dd/mm/yyyy` y valores monetarios que usan coma decimal (ej. `1250,75`).
2. **Inventario (Excel):** archivo `inventario.xlsx` con **tres hojas** (una por ciudad: Guayaquil, Quito, Cuenca), cada una con encabezados que inician en la fila 3 y columnas no siempre idénticas entre hojas.
3. **Reclamos de clientes (JSON anidado):** archivo `reclamos.json` obtenido de una API interna, con estructura anidada (cliente → lista de reclamos → detalle de cada reclamo) que debe ser normalizada a una tabla plana.
4. **Histórico de transacciones (archivo comprimido de gran tamaño):** archivo `historico_2023_2025.csv.gz` que no debe cargarse completo en memoria, sino leerse por **chunks**, filtrando únicamente las transacciones con `monto > 500` a medida que se procesa.

**Requerimientos técnicos**

- Leer correctamente cada fuente respetando su codificación, delimitador y estructura particular, documentando en el notebook cada decisión tomada (encoding, separador, engine, `sheet_name`, `orient`, etc.).
- Detectar y manejar filas corruptas o incompletas durante la lectura (registros con columnas faltantes, tipos de dato inconsistentes o filas vacías), reportando cuántas fueron descartadas y por qué.
- Unificar el inventario de las tres hojas de Excel en un único DataFrame, homologando nombres de columnas.
- Aplanar (normalizar) el JSON de reclamos en una tabla con una fila por reclamo individual.
- Procesar el histórico comprimido por chunks (tamaño de chunk a definir y justificar), acumulando únicamente los registros filtrados sin cargar el archivo completo en memoria.
- Consolidar las cuatro fuentes en un modelo final coherente (al menos una tabla de hechos y su relación con inventario).
- **Escribir los resultados** en al menos tres formatos distintos: (a) un archivo Parquet particionado por ciudad, (b) un archivo Excel con múltiples hojas (resumen ejecutivo, detalle de ventas, reclamos), y (c) un respaldo comprimido en JSON.
- Incluir manejo de excepciones (`try/except`) en cada operación de lectura/escritura, con mensajes informativos.

**Entregables**

Notebook `.ipynb` con: introducción del caso, lectura de las 4 fuentes con su justificación técnica, limpieza y validación, consolidación, escritura en los 3 formatos solicitados, y una sección final de conclusiones donde el estudiante argumente qué formato de salida es más adecuado para cada consumidor de la información (BI, auditoría, respaldo).
