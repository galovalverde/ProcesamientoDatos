## Caso: Auditoría de Calidad de Datos y EDA para la plataforma de e-learning "SabiosApp"

**Contexto**

SabiosApp es una plataforma de cursos en línea que ha exportado un archivo `actividad_estudiantes.csv` con el registro de inscripciones, avance y calificaciones de sus usuarios durante el último semestre. El equipo de datos sospecha que el archivo tiene problemas de calidad que podrían distorsionar cualquier análisis posterior, por lo que le solicitan primero **auditar y limpiar** la información, y luego realizar un **análisis exploratorio** que permita entender el comportamiento de los estudiantes en la plataforma.

**Descripción del dataset**

El archivo contiene, entre otras, las columnas: `id_estudiante`, `curso`, `fecha_inscripcion`, `horas_estudio`, `porcentaje_avance`, `calificacion_final`, `pais` y `dispositivo`. El dataset presenta de forma intencional: registros duplicados, valores faltantes en varias columnas, categorías escritas de forma inconsistente (ej. `"Ecuador"`, `"ecuador"`, `"ECU"`), valores atípicos (outliers) en `horas_estudio` y `calificacion_final`, y columnas con tipos de dato incorrectos (fechas como texto, porcentajes como texto con símbolo `%`).

**Parte 1 — Calidad de Datos**

- Cargar el dataset e identificar y cuantificar: duplicados, valores nulos por columna, categorías inconsistentes y outliers.
- Corregir los tipos de dato (fechas, porcentajes, numéricos).
- Estandarizar las categorías inconsistentes (por ejemplo, unificar el nombre de país).
- Tratar los valores faltantes justificando la estrategia elegida (eliminación, imputación por media/mediana/moda, u otra).
- Tratar los outliers detectados, justificando el criterio utilizado (rango intercuartílico, z-score, u otro).
- Dejar un dataset limpio y documentar, en una celda de resumen, cuántos registros/columnas fueron afectados por cada corrección.

**Parte 2 — Análisis Exploratorio de Datos (EDA)**

- Calcular estadísticas descriptivas generales (media, mediana, desviación estándar, mínimos y máximos) de las variables numéricas.
- Analizar la distribución de `calificacion_final` y `horas_estudio` mediante histogramas y/o boxplots.
- Explorar la relación entre `horas_estudio` y `calificacion_final` (gráfico de dispersión y coeficiente de correlación).
- Comparar el `porcentaje_avance` promedio por curso y por país mediante tablas agrupadas y gráficos de barras.
- Identificar el dispositivo más utilizado y si existe relación entre el tipo de dispositivo y el rendimiento académico.

**Entregables**

Notebook `.ipynb` con: carga de datos, diagnóstico y limpieza documentada (Parte 1), análisis exploratorio con al menos 4 visualizaciones (Parte 2), y una sección final de conclusiones con los hallazgos más relevantes sobre el comportamiento de los estudiantes.

**Criterios de evaluación**

- Correcta identificación y tratamiento de problemas de calidad de datos (35%)
- Justificación de las decisiones de limpieza (imputación, outliers, estandarización) (15%)
- Profundidad y pertinencia del análisis exploratorio (30%)
- Calidad de las visualizaciones (10%)
- Claridad del notebook y conclusiones (10%)
