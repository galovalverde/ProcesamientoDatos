## Caso: Proceso ETL para Recursos Humanos de "Grupo Empresarial Andino"

**Contexto**

Grupo Empresarial Andino es una corporación con varias empresas afiliadas que hasta ahora ha manejado la información de su personal en archivos sueltos (Excel y CSV) actualizados manualmente cada mes. El área de Recursos Humanos necesita reportes confiables sobre nómina, departamentos y desempeño, por lo que le solicitan diseñar e implementar un **proceso ETL completo**: extraer la información desde sus fuentes originales, transformarla y limpiarla, cargarla en una base de datos relacional (SQLite), y finalmente construir las consultas SQL que respondan a las preguntas del negocio.

**Fuentes de datos**

1. **`empleados.csv`**: datos maestros de cada empleado (`id_empleado`, `nombre`, `id_departamento`, `fecha_ingreso`, `cargo`, `salario_base`).
2. **`departamentos.xlsx`**: catálogo de departamentos (`id_departamento`, `nombre_departamento`, `presupuesto_anual`).
3. **`evaluaciones.csv`**: evaluaciones de desempeño trimestrales (`id_empleado`, `trimestre`, `calificacion_desempeno`, `bono_asignado`), con varias filas por empleado.

**Parte 1 — Diseño del proceso ETL (Extract)**

- Diseñar (puede ser un diagrama simple en una celda markdown, o una descripción estructurada) las etapas del proceso ETL que va a implementar: qué se extrae, qué transformaciones se aplican y en qué orden, y cómo se carga el resultado.
- Implementar la etapa de **extracción**: cargar las tres fuentes verificando su estructura antes de continuar (número de columnas esperado, tipos de dato básicos).

**Parte 2 — Transformación (Transform)**

- Limpiar y estandarizar los datos de las tres fuentes (nulos, formatos de fecha, nombres de columnas).
- Calcular al menos dos columnas derivadas útiles para el negocio (por ejemplo, antigüedad en años a partir de `fecha_ingreso`, o promedio de `calificacion_desempeno` por empleado a lo largo de los trimestres).
- Dejar las tres tablas transformadas y listas para ser cargadas, respetando el modelo relacional (claves primarias/foráneas coherentes entre `empleados`, `departamentos` y `evaluaciones`).

**Parte 3 — Carga (Load) a una base de datos SQL**

- Crear una base de datos SQLite y cargar en ella las tres tablas transformadas, definiendo explícitamente los tipos de dato de cada columna.
- Verificar la carga contando registros por tabla y comprobando que no se perdió información en el proceso.

**Parte 4 — Consultas SQL para responder preguntas del negocio**

Usando exclusivamente sentencias SQL sobre la base de datos cargada (con `sqlite3` o `pandas.read_sql`), responder:

- ¿Cuál es el salario promedio por departamento?
- ¿Qué empleados tienen una calificación de desempeño promedio superior a 8 en los últimos trimestres disponibles?
- ¿Qué departamento ha asignado el mayor monto total de bonos?
- ¿Qué porcentaje del presupuesto anual de cada departamento representa la suma de salarios de sus empleados?

Cada respuesta debe obtenerse con al menos una consulta que use `JOIN` y una que use `GROUP BY`.

**Entregables**

Notebook `.ipynb` con: diseño del proceso ETL (Parte 1), extracción y transformación documentadas (Parte 1-2), carga a SQLite verificada (Parte 3), las cuatro consultas SQL con su resultado interpretado (Parte 4), y una sección final de conclusiones sobre qué automatizaría primero si este proceso se ejecutara mensualmente.
